 # Data Analysis with Pandas: Comprehensive Guide



 This notebook serves as an educational reference for fundamental concepts and operations in the **Pandas** library.



 ---



 ## 1. Environment Setup

 To begin, we import the Pandas library. By convention, it is aliased as `pd`.

```python
import pandas as pd
```

 ## 2. Creating DataFrames from Raw Data

 A Pandas DataFrame can be constructed from various raw Python data structures, such as lists of lists or lists of dictionaries.



 ### Method A: Constructing from a List of Lists

 We define a 2D matrix (list of lists) containing observation rows, and a separate list specifying column labels.

```python
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
```

 We initialize the `pd.DataFrame` constructor by passing both the data matrix and explicit column names.

```python
df = pd.DataFrame(data, columns=columns)
```

 ### Method B: Constructing from a List of Dictionaries (Records)

 Alternatively, data often comes as a list of dictionaries (JSON/Records format). Here, each dictionary represents a row, where keys correspond to column names.

```python
data = [
    {
        "Make": "Nissan",
        "Model": "Stanza",
        "Year": 1991,
        "Engine HP": 138.0,
        "Engine Cylinders": 4,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "sedan",
        "MSRP": 2000
    },
    {
        "Make": "Hyundai",
        "Model": "Sonata",
        "Year": 2017,
        "Engine HP": None,
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "Sedan",
        "MSRP": 27150
    },
    {
        "Make": "Lotus",
        "Model": "Elise",
        "Year": 2010,
        "Engine HP": 218.0,
        "Engine Cylinders": 4,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "convertible",
        "MSRP": 54990
    },
    {
        "Make": "GMC",
        "Model": "Acadia",
        "Year": 2017,
        "Engine HP": 194.0,
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "4dr SUV",
        "MSRP": 34450
    },
    {
        "Make": "Nissan",
        "Model": "Frontier",
        "Year": 2017,
        "Engine HP": 261.0,
        "Engine Cylinders": 6,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "Pickup",
        "MSRP": 32340
    }
]
```

 When instantiating a DataFrame from a list of dictionaries, Pandas automatically infers column names from the dictionary keys.

```python
df = pd.DataFrame(data)
```

 ## 3. Inspecting and Accessing Data



 ### Visualizing Top Rows (`head`)

 The `.head(n)` method displays the first $n$ rows of the DataFrame (default is $n = 5$). This provides a quick inspection of the dataset structure.

```python
df.head(3)
```

 ### Accessing Columns (Series)

 Each individual column in a DataFrame is represented as a **Pandas Series**.

 You can access a column using:

 1. **Dot notation**: `df.ColumnName` (Only works if column names contain no spaces or special characters).

 2. **Bracket notation**: `df['Column Name']` (Preferred and robust for all string names).

```python
df.Make
df["Engine HP"]
```

 ### Subsetting DataFrames (Column Projection)

 To extract a subset of columns as a new DataFrame, pass a Python **list of column names** inside the indexing operator: `df[['Col1', 'Col2']]`.

```python
df[["Make", "Model", "MSRP"]]
```

 ## 4. Feature Engineering and Column Manipulations



 ### Vectorized Column Arithmetic

 We can add new columns by applying element-wise operations directly across existing Series.

```python
df['Shamsi Year'] = df['Year'] - 621
df
```

 ### Applying Custom Functions (`.apply` with `lambda`)

 For non-vectorized transformations or logical conditions, the `.apply()` method executes a function across every element in a Series.

```python
df['is new'] = df['Year'].apply(lambda x: "Yes" if x > 2011 else "No")
df
```

 ### Manual Assignment of Sequences

 New columns can also be created by directly assigning a Python list, provided its length matches the DataFrame length.

```python
df['id'] = [1, 2, 3, 4, 5]
df
```

 ### Dropping Columns (`del` vs. `.drop()`)

 Column deletion can be done using:

 - `del df['col_name']`: Modifies the DataFrame in-place directly.

 - `df.drop(columns=[...])`: Offers flexibility (dropping multiple columns at once). Using `inplace=True` alters the original object directly.

```python
del df['id']
df.drop(columns=['Shamsi Year', 'is new'], inplace=True)
```

 ## 5. Indexing and Row Selection



 ### Label-Based Indexing (`.loc`)

 Every DataFrame row possesses an index identifier. The `.loc[]` property retrieves rows by their explicit **index label**.

```python
df.loc[[1, 2]]
```

 ### Modifying Index Labels

 Indexes do not need to be numerical; we can reassign custom labels (e.g., categorical strings).

```python
df.index = ['A', 'B', 'C', 'D', 'E']
df
```

 ### Positional Indexing (`.iloc`)

 Regardless of index labels (custom or non-numeric), `.iloc[]` retrieves rows based on their **integer positional index** (0-indexed).

```python
df.iloc[[1, 2]]
```

 ### Resetting DataFrame Index

 The `.reset_index()` method restores index values back to sequential integers starting at 0.

 - By default, the old index is retained as a new column. Setting `drop=True` removes it completely.

 - Setting `inplace=True` persists changes without returning a new DataFrame.

```python
df.reset_index(drop=True, inplace=True)
```

 ## 6. Element-wise Operations & Logical Filtering



 ### Vectorized Operations

 Like NumPy, mathematical operations applied on Series propagate across every individual element automatically.

```python
df['Engine HP'] / 100           
```

 ### Vectorized Comparisons

 Logical conditions generate a Series of boolean values (`True`/`False`).

```python
df['Year'] >= 2015
```

 ### Boolean Indexing (Filtering)

 Passing a boolean Series into a DataFrame filters and returns only rows where the boolean expression resolves to `True`.

```python
df[df['Make'] == "Nissan"]
```

 ### Compound Boolean Filtering

 Multiple conditions can be combined using bitwise logical operators:

 - `&` for **AND** logic

 - `|` for **OR** logic



 *Note: Each condition must be enclosed in parentheses `()` due to operator precedence in Python.*

```python
df[(df['Make'] == "Nissan") | (df['Year'] >= 2015)]
```

 ## 7. String Manipulations (`.str` Accessor)



 String operations on text columns are accessed using the `.str` namespace accessor.



 ### Capitalization

 Converts text values into capitalized standard format.

```python
df['Vehicle_Style'].str.capitalize()
```

 ### String Replacement

 Replaces specific substring characters (e.g., replacing spaces with underscores).

```python
df['Vehicle_Style'].str.replace(pat=" ", repl="_")
```

 ### Method Chaining

 Multiple string methods can be chained sequentially for complex textual cleanups.

```python
df['Vehicle_Style'].str.capitalize().str.replace(pat=" ", repl="_")
```

 ### Persisting Changes

 Reassigning the transformed Series back to the column updates the original DataFrame.

```python
df['Vehicle_Style'] = df['Vehicle_Style'].str.capitalize().str.replace(pat=" ", repl="_")
df
```

 ## 8. Summary Statistics and Descriptive Analytics



 ### Individual Summary Aggregations

 Aggregation functions summarize a numeric Series into a single scalar value.

```python
print(df.MSRP.min())
print(df.MSRP.max())
print(df.MSRP.sum())
print(df.MSRP.mean())
print(df.MSRP.median())
print(df.MSRP.std())
print(df.MSRP.var())
```

 ### Series Descriptive Statistics (`.describe()`)

 The `.describe()` method returns a standard summary of basic statistical indicators for a Series.

```python
df.MSRP.describe()
```

 ### DataFrame Descriptive Statistics

 Applied to an entire DataFrame, `.describe()` automatically selects all numeric columns and returns a tabular statistical overview.

```python
df.describe()
```

 ### Formatting Outputs

 Analytical results can be formatted (e.g., controlling decimal precision) using numerical methods like `.round()`.

```python
df.describe().round(2)
```

 ## 9. Categorical and Missing Data Analysis



 ### Unique Values Inspection

 The `.unique()` method lists all distinct elements contained within a Series.

```python
df.Make.unique()
```

 ### Unique Values Count

 The `.nunique()` method counts the number of distinct values in a column (excluding missing entries).

```python
df.Make.nunique()
```

 ### Missing Value Detection (`.isnull()`)

 Returns a boolean mask indicating `True` where values are missing (`NaN` or `None`).

```python
df.isnull()
```

 ### Summarizing Missing Values

 Combining `.isnull()` with `.sum()` aggregates missing counts for each column, providing a quick data completeness audit.

```python
df.isnull().sum()
```

 ## 10. Data Aggregation (`groupby`)



 The `.groupby()` method implements the standard **Split-Apply-Combine** workflow:

 1. **Split**: Groups the DataFrame by distinct values of specified column(s).

 2. **Apply**: Computes a summary statistic (e.g., `.mean()`) on selected variables (`MSRP`).

 3. **Combine**: Merges the aggregated metrics into a structured summary.

```python
df.groupby("Transmission Type").MSRP.mean().round(2)
```

 ## 11. Interoperability with NumPy



 Pandas is built on top of NumPy arrays. Extracting the raw underlying array structure from a Series can be achieved via the `.values` attribute (or `.to_numpy()`).

```python
np_array_year = df['Year'].values
print(np_array_year) 
```
