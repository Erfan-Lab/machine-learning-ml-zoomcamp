```python
data = [
    {
        "Make": "Nissan",
        "Model": "Stanza",
        "Year": 1991,
        "Engine HP": 138.0,
        "Engine Cylinders": 4,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "sedan",
        "MSRP": 2000
    },
    {
        "Make": "Hyundai",
        "Model": "Sonata",
        "Year": 2017,
        "Engine HP": None,
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "Sedan",
        "MSRP": 27150
    },
    {
        "Make": "Lotus",
        "Model": "Elise",
        "Year": 2010,
        "Engine HP": 218.0,
        "Engine Cylinders": 4,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "convertible",
        "MSRP": 54990
    },
    {
        "Make": "GMC",
        "Model": "Acadia",
        "Year": 2017,
        "Engine HP": 194.0,
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "4dr SUV",
        "MSRP": 34450
    },
    {
        "Make": "Nissan",
        "Model": "Frontier",
        "Year": 2017,
        "Engine HP": 261.0,
        "Engine Cylinders": 6,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "Pickup",
        "MSRP": 32340
    }
]
```

```python
df = pd.DataFrame(data)
```

```python
df.head(3) # returns the first n rows, n = 5 is the default.
```

```python
# every column of a dataframe is a series. 
df.Make
df["Engine HP"]
```

```python
# access a subset of the dataframe:
df[["Make", "Model", "MSRP"]]
```

```python
# add a column to a dataframe: 
df['Shamsi Year'] = df['Year'] - 621
df
```

```python
# apply method for adding a new column with a function to determine the values:
df['is new'] = df['Year'].apply(lambda x: "Yes" if x > 2011 else "No")
df
```

```python
# add a new column manually: 
df['id'] = [1, 2, 3, 4, 5]
df
```

```python
# deleting a column / multiple columns from a dataframe:
del df['id']
df.drop(columns=['Shamsi Year', 'is new'], inplace=True)
```

```python
# each row of a dataframe has an index.
# we access the rows using the index, using the loc method. 
df.loc[[1, 2]]
```

```python
# when creating a dataframe, we can explicitly determine the indices if we don't want it to be the default (a range from 0 to n-1).
df.index = ['A', 'B', 'C', 'D', 'E']
df
```

```python
# even after changing the default indices, we can access the rows with their positional index using the iloc method:
df.iloc[[1, 2]]
```

```python
# to get back to the initial indices, we use the reset_index method:
df.reset_index()
# this keeps the previous values as a column called index. if we don't want it, we use drop=True.
df.reset_index(drop=True)
# this method does not change the dataframe. if we want it to do so, we use inplace=True.
df.reset_index(drop=True, inplace=True)
```

```python
# as in numpy arrays, we can do element-wise operations on columns of a pandas dataframe. 
df['Engine HP'] / 100           
```

```python
# we can also do comparisons as we did in numpy arrays.
df['Year'] >= 2015
```

```python
# we can use the comparisons (or more percisely, the logical operators), to filter the data in a dataframe.
df[df['Make'] == "Nissan"]
```

```python
# we can also combine our conditions with and/or (&, |):
df[(df['Make'] == "Nissan") | (df['Year'] >= 2015)]
```

```python
# pandas supports string operations unlike numpy (numpy is written for numeric calculations).
# to use string methods on pandas columns, we use the .str method:
df['Vehicle_Style'].str.capitalize()
```

```python
# we can also use the replace method to transform any space characters to underscores:
df['Vehicle_Style'].str.replace(pat=" ", repl="_")
```

```python
# we can also chain these operations to do the changes all together:
df['Vehicle_Style'].str.capitalize().str.replace(pat=" ", repl="_")
```

```python
# to save these changes, we can assign the new values to the dataframe.
df['Vehicle_Style'] = df['Vehicle_Style'].str.capitalize().str.replace(pat=" ", repl="_")
df
```

```python
# as in numpy, we have the summarizing operators(min, max, sum, etc.):
print(df.MSRP.min())
print(df.MSRP.max())
print(df.MSRP.sum())
print(df.MSRP.mean())
print(df.MSRP.median())
print(df.MSRP.std())
print(df.MSRP.var())
```

```python
# we also have the describe method which provides all the useful information all in one:
df.MSRP.describe()
```

```python
# we can also call the describe method on the whole dataframe (calcualte the same information for numeric columns):
df.describe()
```

```python
# we can use the round function to show the numbers rounded untill n decimal points. 
df.describe().round(2)
```

```python
# see the set of values of a column:
df.Make.unique()
```

```python
# see the number of unique values in a column:
df.Make.nunique()
```

```python
# to see the null values in the dataframe, we use the isnull() method:
df.isnull()
```

```python
# we can use the sum method alongside the isnull method to find out the number of null values in each column:
df.isnull().sum()
```

```python
# we can group the data of the dataframe by the values of a certain column using the groupby method:
df.groupby("Transmission Type").MSRP.mean().round(2)
```

```python
# we can access the numpy array of the values using the values method.
np_array_year = df['Year'].values
print(np_array_year)
```
