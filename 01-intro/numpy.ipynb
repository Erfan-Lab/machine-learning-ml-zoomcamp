# Introduction to NumPy
This notebook introduces the NumPy library and demonstrates common array creation, indexing, random number generation, arithmetic, and linear algebra operations. **All code cells are left unchanged.**

```python
import numpy as np
```

## Creating an Array of Zeros
`np.zeros(shape)` creates a new NumPy array filled entirely with zeros. Here, a one-dimensional array of length **5** is created.

```python
np.zeros(5)
```

## Creating an Array of Ones
`np.ones(shape)` returns an array whose elements are all initialized to **1**.

```python
np.ones(5)
```

## Filling an Array with a Constant Value
`np.full(shape, fill_value)` creates an array where every element has the specified value. This example fills an array of length **10** with `2.5`.

```python
np.full(10, 2.5)
```

## Creating Arrays from Python Lists
`np.array()` converts standard Python lists into efficient NumPy arrays.

```python
arr = np.array([1, 3, 5, 7, 9])
```

## Generating Evenly Spaced Integers
`np.arange(start, stop, step)` creates values from **start** to **stop (exclusive)** using the specified step size.

```python
arr2 = np.arange(3, 10, 2)
```

## Generating Evenly Spaced Numbers
`np.linspace(start, stop, num)` generates a fixed number of evenly spaced values including both endpoints.

```python
np.linspace(0, 10, 11)
```

## Creating a Two-Dimensional Array
Passing a tuple to `np.ones()` creates a matrix. This example creates a **3×3** matrix filled with ones.

```python
np.ones((3, 3))
```

## Creating a Three-Dimensional Array
NumPy supports multidimensional arrays. This example creates a **2×2×2** array filled with zeros.

```python
np.zeros((2, 2, 2))
```

## Building a Matrix
This cell creates a 3×3 matrix that will be used to demonstrate indexing and modification.

```python
arr3 = np.array(([4, 5, 6], [7, 8, 9], [10, 11, 12]))
```

## Modifying a Single Element
Individual elements can be accessed and updated using row and column indices.

```python
arr3[1][1] = 1000
print(arr3)
arr3[1, 1] = 911
print(arr3)
```

## Replacing an Entire Row
Assigning to `arr[row]` replaces every value in that row.

```python
arr3[0] = [1, 1, 1]
arr3
```

## Replacing an Entire Column
The slicing syntax `:` selects all rows, allowing an entire column to be modified at once.

```python
arr3[:, 2] = [0, 0, 0]
arr3
```

## Random Floating-Point Numbers
`np.random.rand()` generates random values sampled from a uniform distribution over the interval [0, 1). Multiplying by 100 scales the range.

```python
np.random.rand(2, 2, 2) * 100
```

## Reproducible Random Numbers
Using `np.random.seed()` fixes the random number generator so the same sequence is produced every time the notebook runs.

```python
# np.random.seed(1) ==> this uses a fixed number for the seed so the output will always be the same. 
np.random.rand(3, 3) 
```

## Random Numbers from Different Distributions
This section compares common NumPy random distributions such as the standard uniform and standard normal distributions.

```python
# the default generation of random numbers is standard uniform distribution. 
# if we want normal distribution, we can use np.random.randn(3, 3) instead.
```

## Random Integers
`np.random.randint(low, high, size)` generates random integers in the specified range and shape.

```python
np.random.randint(0, 10, (2, 3, 4))
```

## Creating a Sample Array
A simple array is created for demonstrating arithmetic operations.

```python
arr4 = np.arange(10)
```

## In-place Arithmetic
NumPy allows arithmetic operations to be applied directly to every element of an array without explicit loops.

```python
arr4 += 1
arr4 = arr4 / 0.5
arr4
```

## Vectorized Expressions
NumPy performs element-wise arithmetic efficiently, allowing complex mathematical expressions to be evaluated on entire arrays.

```python
arr4 = arr4 * 5 + arr4 / 0.25
arr4
```

## Boolean Indexing
Boolean masks allow selecting elements that satisfy specific conditions.

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([5, 4, 3, 2, 1])
print(a > b)
print(a >= b)
print(a < b)
print(a <= b)
print(a == b)
print(a >= 3)
print(b <= 3)
```

## Aggregation Functions
Functions such as `min()`, `max()`, `mean()`, and `sum()` summarize an array's contents.

```python
print(a[a >= b])
print(b[a <= 3])
```

## Statistical Operations
This section demonstrates several built-in statistical methods available for NumPy arrays.

```python
print(a.min())
print(a.max())
print(a.sum())
print(a.mean())
print(a.std())
```

## Dot Product
The dot product is a fundamental linear algebra operation used in machine learning, graphics, and scientific computing.

```python
# dot production using numpy:
# works for vector-vector multiplication and matrix-vector multiplication and matrix-matrix multiplication.
# vector-vector: 
u = np.arange(5)
v = np.array([1, 3, 5, 7, 9])
print(u.dot(v))
U = np.arange(15)
U = U.reshape((3, 5))
print(U.dot(v))
V = np.arange(15)
V += 5
V *= 2
V = V.reshape((5, 3))
print(U.dot(V))
```

## Identity Matrix
`np.eye(n)` creates an identity matrix containing ones on the main diagonal and zeros elsewhere.

```python
# identity matrix using numpy: 
# square matrix mith ones on the main diagonal and zero on other cells.
# to create an n*n identity matrix, we use the eye function.
identity_3 = np.eye(3)
print(identity_3)
print(identity_3.dot(U))
```

## Matrix Inverse
`np.linalg.inv()` computes the inverse of a square matrix when it exists.

```python
# matrix inverse using numpy:
# to calculate the inverse of a matrix using built-in numpy functions, we use linalg.inv function.
matrix1 = np.array([[1, 2], [3, 4]])
print(np.linalg.inv(matrix1))
matrix2 = np.array([[1, 2], [2, 4]])
try:
    print(np.linalg.inv(matrix2))
except:
    print("the matrix does not have an inverse!")
```
